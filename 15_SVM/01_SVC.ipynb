## `(SVC)` Support Vector Classifiers

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import make_classification
```

```python
X,y=make_classification(n_samples=1000,n_features=2,n_classes=2,
                        n_clusters_per_class=2,n_redundant=0)
```

```python
X
```

```python
y
```

```python
df1 = pd.DataFrame(X)[0]
df2 = pd.DataFrame(X)[1]
```

```python
sns.scatterplot(x=(pd.DataFrame(X)[0]), y=(pd.DataFrame(X)[1]), hue=y)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size = 0.2,random_state=10)
```

```python
from sklearn.svm import SVC
```

```python
svc = SVC(kernel='linear')
```

```python
svc.fit(X_train,y_train)
```

```python
## Prediction
y_pred = svc.predict(X_test)
```

```python
from sklearn.metrics import classification_report,confusion_matrix
```

```python
print(confusion_matrix(y_test,y_pred))
print(classification_report(y_test,y_pred))
```

### Using RBF kernel

```python
rbf = SVC(kernel='rbf')
```

```python
rbf.fit(X_train,y_train)
```

```python
## Prediction
y_pred1 = svc.predict(X_test)
```

```python
print(confusion_matrix(y_test,y_pred1))
print(classification_report(y_test,y_pred1))
```

### Using polynomial kernel 

```python
polynomial = SVC(kernel='poly')
polynomial.fit(X_train,y_train)
```

```python
y_pred3 = polynomial.predict(X_test)
```

```python
print(confusion_matrix(y_test,y_pred3))
print(classification_report(y_test,y_pred3))
```

### Sigmoid Kernal

```python
sigmoid = SVC(kernel='sigmoid')
sigmoid.fit(X_train,y_train)
```

```python
y_pred4 = sigmoid.predict(X_test)
print(confusion_matrix(y_test,y_pred4))
print(classification_report(y_test,y_pred4))
```

## Hyperparameter tuning

```python
from sklearn.model_selection import GridSearchCV

# Defining the parameter grid for hyperparameter tuning
param_grid = {
    'C': [0.1, 1, 10, 100],
    'gamma': [0.001, 0.01, 0.1, 1],
    'kernel': ['rbf']
}
```

```python

grid = GridSearchCV(SVC(), param_grid, cv=5, refit=True, verbose=3)
```

```python
grid.fit(X_train, y_train)
```

```python
grid.best_params_
```

```python
# Predicting with the best estimator from GridSearchCV
y_pred4 = grid.predict(X_test)
print(confusion_matrix(y_test,y_pred4))
print(classification_report(y_test,y_pred4))
```
